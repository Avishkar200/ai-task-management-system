```python
from pathlib import Path
import sys
p = Path(r'C:/Users/ASHISH S YADAV.LAPTOP-LEK06VOJ/Downloads/task.pdf')
print('exists', p.exists(), 'size', p.stat().st_size if p.exists() else 'missing')
for mod in ['pymupdf', 'fitz', 'pypdf', 'PyMuPDF']:
    try:
        __import__(mod)
        print('module_ok', mod)
    except Exception as e:
        print('module_err', mod, type(e).__name__, e)
try:
    import fitz
    doc = fitz.open(str(p))
    text = '\n'.join(page.get_text() for page in doc)
    print('TEXT_START')
    print(text[:5000])
    print('TEXT_END')
except Exception as e:
    print('FITZ_ERROR', type(e).__name__, e)
try:
    from pypdf import PdfReader
    reader = PdfReader(str(p))
    text = '\n'.join(page.extract_text() or '' for page in reader.pages)
    print('PDF_READER_TEXT_START')
    print(text[:5000])
    print('PDF_READER_TEXT_END')
except Exception as e:
    print('PDF_READER_ERROR', type(e).__name__, e)
```
